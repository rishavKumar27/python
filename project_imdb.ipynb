# Q3. How do we clean the `duration` column? (e.g., `'175 min'` → `175`)

```python
df['duration']
```

```python
# df["duration"].astype(int)
#conveting into int but 'min' is string so 1st have to replace it with nothing & then covert it in int
```

```python
df["duration"] = df['duration'].str.replace("min","")
```

#Q. Convert gross_income and votes into Int

Q4. How do we clean the `gross` column and `votes` column?


```python
df["gross_income"]
```

```python
#Step 1 : CHECK : df["gross_income"].astype(float)

# By doing this we'll know whats the error coming that we have to replace

# While conversion '$', 'M', ',' are coming in the data so we cannot convert it into float

# So first we have to replace all these things with nothing
```

```python
df["gross_income"] = df["gross_income"].str.replace(",","")
df["gross_income"] = df["gross_income"].str.replace("$","")
df["gross_income"] = df["gross_income"].str.replace("M","")
```

```python
df["gross_income"] = df["gross_income"].astype(float)
df["gross_income"]
```

```python
#VOTES into INT
df["votes"] = df["votes"].str.replace(",", "")
```

```python
df["votes"] = df["votes"].astype(int)
```

```python
df.info()
```

# Q22. Which Director + Actor duo is the most frequent?

```python
# 1. Create a simplified copy
duo_df = df[['directors_name', 'stars_name']].copy()

# 2. Split on comma ',' first, then explode both columns
duo_df['directors_name'] = duo_df['directors_name'].str.split(',')
duo_df = duo_df.explode('directors_name')

duo_df['stars_name'] = duo_df['stars_name'].str.split(',')
duo_df = duo_df.explode('stars_name')

# 3. Clean any remaining whitespaces from the individual names
duo_df['directors_name'] = duo_df['directors_name'].str.strip()
duo_df['stars_name'] = duo_df['stars_name'].str.strip()

# 4. Combine them and find the most frequent duo
(duo_df['directors_name'] + " & " + duo_df['stars_name']).value_counts().head(5)
```

### Q1. Plot the average movie rating over the years (Line Chart)

```python
roy = df.groupby("year")["rating"].mean().sort_index()
# roy = rate over year
```

```python
plt.figure(figsize=(12,6))

plt.plot(roy.index, roy.values, color = "purple", marker = "*", markersize = 5, linestyle = "-.")

plt.title("Average Movie Rating Over the Years")
plt.xlabel("Years", color = "darkred")
plt.ylabel("Avgerage Rating")
plt.grid(alpha=0.4)
plt.show()
```

# Q3. Plot the distribution of movie ratings (Histogram using Matplotlib)

```python
plt.figure(figsize=(14,6))

plt.hist(df["rating"] , color = 'pink', edgecolor = "black", bins = 30)

plt.title("Distribution Of Movie Rating")
plt.xlabel("Ratings", color = "darkred")
plt.ylabel("Frequency")
plt.show()
```

# Q5. Which are the top 10 most frequent actors? (Horizontal Bar Chart)

```python
# Split by comma, explode to get individual actor rows, and strip whitespace
x = df['stars_name'].str.split(',').explode()
top_10 = x.value_counts().sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(13,6))

plt.bar(top_10.index, top_10.values, color = ['pink','aqua','skyblue'], edgecolor = 'black')
#to make this chart look horizontal use (barh) instead of (bar)
plt.title("Top 10 Most Frequent Actors")
plt.xticks(rotation = 45)
plt.show()
```

# Q6. Plot the relationship between year and rating (Scatter Plot)
