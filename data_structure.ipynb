# **LIST**

```python
lst = ["rishav", 23,True, 5.11, 'B']
print(lst)
#accessing through index
print(lst[1:3])

#modify values
lst[1] = 21
print(lst)
#slicing
a= ["rishav", "simran", "ankur","shreya","xotic",21,True]
print(a[1:3])
```

```python
#reversing a string
a= ["rishav", "simran", "ankur","shreya","xotic",21,True]
print(a[::-1])
print('-'*20)

#appending
a.append("shanu")
print(a)
print('-'*20)

#removing
a.remove("shreya")
print(a)
print('-'*20)
```

```python
#sorting
a=[1,2,5,7,8,5,4,51,48,325,45,62,8,6,5]
print(a)
print(sorted(a))
print('-'*20)

#finding element
a= ["rishav", "simran", "ankur","shreya","xotic",21,True]
print(a.index('simran'))
print('-'*20)

#countng
print(a.count('rishav'))
print('-'*20)

#extending the list
a.extend(['rohan','amit','sahil'])
print(a)
print('-'*20)

#maximum and minimum in the list
g=[1,2,5,7,8,5,4,51,48,325,45,62,8,6,5]
print(min(g))
print(max(g))
print('-'*20)

#accessing/iterating the elememts of the list  | Direct
lst=['rishav', 'simran', 'ankur', 'shreya', 'xotic', 21, True, 'rohan', 'amit', 'sahil']
for i in lst:
  print(i)

print('-'*20)
#accessing/iterating the elememts of the list  | loop
print(lst)
for i in range(len(lst)):
  print(i,lst[i])

print('-'*20)
#Iterating the elememts of the list in reverse
for i in range(len(lst)-1,-1,-1):
  print(i,lst[i])

```

**MULTI DIMENSIONAL LIST**

```python
lst= [['rishav',21,'bhopal',True],
 ['simran',21,'bhopal',True],
 [1,2,3,[4,5,6]]]
#accessing
print(lst)
print(lst[0])
print(lst[0][0])
print(lst[1])
print(lst[1][0])
print(lst[2])
print('-'*20)

#modifying
lst[0][1]=23
print(lst)
print('-'*20)

#appending
lst.append(['rohan',22,'jamshedpur'])
print(lst)
print('-'*20)

#deleting the index
del lst[2]
print(lst)
print('-'*20)

```

```python
#accessing all the elements

lst = [['rishav', 23, 'bhopal', True], ['simran', 21, 'bhopal', True], ['rohan', 22, 'jamshedpur']]
for i in lst:
  print(i)
```

### Iterating through sub-elements of a multi-dimensional list

```python
# Iterating through each sub-element within the sublists

lst = [['rishav', 23, 'bhopal', True], ['simran', 21, 'bhopal', True], ['rohan', 22, 'jamshedpur']]

print('Iterating through all sub-elements:')
for sublist in lst:
  for item in sublist:
    print(item)
```

# **DICTIONARY**

```python
dct = {1 : 'rishav', 2 : 'simran', 3 : 'vikas', 4 : 'sakshi', 5 : 'riya'}
print(dct)
print(dct[4])
print('-'*50)

#adding and deleting
dct[6] = 'amit'
print(dct)

del dct[5]
print(dct)
print('-'*50)

dct.clear()
print(dct)

#iterating through the dictionary
dct = {1 : 'rishav', 2 : 'simran', 3 : 'vikas', 4 : 'sakshi', 5 : 'riya'}
print(dct.keys())
print(dct.values())

for i in dct.keys():
  print(i, dct[i])

print('-'*20)
print(dct.items())

for k,v in dct.items():
  print(k,v)

print('-'*20)
#Check if key is present or not
print(dct)
print(1 in dct)
print(2 in dct)
print(7 in dct)
print('-'*20)

#updating in the dictioaries

dct_1 = {1 : 'hi', 2 : 'hola', 3 : 'hello'}
dct_2 = {1 : 'yoo', 2 : 'wassup', 3 : 'supp'}
print(dct_1)
print(dct_2)
dct_1.update(dct_2)
print(dct_1)
print(dct_2)
print('-'*50)

```

# **Multi-dimensional Dictionary**

```python
dct = {1 : {'name' : 'rishav', 'age' : 22, 'phone' : 9864512},
      2 : {'name' : 'ankur', 'age' : 20, 'phone' : 785654},
      3 : {'name' : 'simm', 'age' : 21, 'phone' :156586 }}
print(dct)
print('-'*50)

#accessing the elements
print(dct[3])
print(dct[1]['name'])
print('-'*50)

#adding, updating & deleting
print(dct)
dct[4] = {'name' : 'diksha', 'age' : 23, 'phone' :654785 }
print(dct)

print('-'*20)
dct[2]['name'] = 'rohan'
print(dct[2])

print('-'*20)

print(dct)
del dct[4]['age']
print(dct)
```

```python
#Going through the data/loop
dct = { 1 : {'name' : 'rishav', 'age' : 22, 'phone' : 9864512},
        2 : {'name' : 'ankur',  'age' : 20, 'phone' : 785654},
        3 : {'name' : 'simm',   'age' : 21, 'phone' :156586 }}

for i in dct.keys():
  print(i,dct[i]['name'],dct[i]['phone'])

print(2 in dct)
print('-'*50)

data = { 1 : {'name' : 'rishav', 'age' : 22, 'phone' :9864512,'marks' : {'hin':43,'math':46,'sci':42}},
         2 : {'name' : 'ankur',  'age' : 20, 'phone' :785654, 'marks' : {'hin':41,'math':26,'sci':32}},
         3 : {'name' : 'simm',   'age' : 21, 'phone' :156586, 'marks' : {'hin':49,'math':36,'sci':47} }}

for i in data.keys():
  print(i,data[i]['name'],data[i]['marks'])
```

# Dictionary Comprehension

```python
dct = {i:i**2 for i in range(1,6)}
print(dct)
print('-'*35)

#dictionary with condition
dct = {i:i**2 for i in range(1,15) if i%2==0}
print(dct)
print('-'*55)

#Dictionaries from lists
lst = ['apple','banana','kiwi']
dct = {items:len(items) for items in lst}
print(dct)

dct = {len(items):items for items in lst}
print(dct)
print('-'*35)

```
