**OPERATIONS**

```python
# addition
print(2+4)
print(2.3 +5)
print(True + False)
print( 'hey ' + 'there' + ' im learning python')
```

```python
# subtraction
print(2-4)
print(False-True)
print(4.5-3)
```

```python
#multiplication
print(2*4)
print(False*True)
print(4.5*3)
print(2.4*5.7)
```

```python
#division
print(2/4)
print(False/True)
print(4.5/3)
print(2.4/5.7)
print("------------------------")
#floor division (quotient)

print(5//4)
print(3//2.4)
print(2.3//7)
```

```python
#modulo (remainder)
print(2%4)
print(False%True)
print(4.5%3)
print(2.4%5.7)
print(7%3)
print("------------------------")

#exponential(to the power of)
print(2**3)
print(4.5**3)
print(2.4**5.7)
print(7**3)
```

```python
#abs(negative to positive)
print(abs(-43))
print(abs(-4.353))
print("------------------------")

#round
print(round(4.765456))
print(round(2.9344567,2))

print("------------------------")
#pow
print(pow(2,3))
print(pow(7,9))
```

```python
#Comparison Operator
a = 13
b = 33

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
#Logical Operator
#AND
print(True and True)
print(True and False)
print(False and False)
print(False and True)

#OR
print(True or True)
print(True or False)
print(False or False)
print(False or True)

#NOT
print(not True)
print(not False)
```

# **Strings**

```python
print('rishav'+" "+'gupta')
print('simm'*3)
print('-'*20)

#String length
print(len('rishav'))
print("________________________")



#String Slicing
print('rishav gupta'[4])
print('rishav'[-4])
print('rishav gupta'[0:6])
print('rishav gupta'[6:12])
print('rishav gupta'[2:12])
print('rishav gupta'[-1:-8])
print("________________________")

#String Case Conversion
print("rishabh".upper())
print('RISHABH'.lower())
print("________________________")


#String Stripping
print('    rishav    gupta    ')
print('             rishav gupta   '.strip())
print("________________________")

#String Replacing
print('rishav gupta'.replace('a','$'))
print("________________________")

#String Count
print("missippippi".count('p'))
print("________________________")

#String Find
print('rishav gupta'.find('rsh'))
```

```python
# String Reversing using Slicing
my_string = "hello world"
reversed_string = my_string[::-1]
print(f"Original string: {my_string}")
print(f"Reversed string: {reversed_string}")
```

```python
#calculating the probablity of number when two dice thrown at once.
for n in range(1,13):
  number=0
  for i in range(1,7):
    for j in range(1,7):

        if(i+j==n):
          number+=1

  print(n,"=",round((number/36)*100, 2))
```

```python
#calculating the probablity of number when three dice thrown at once.

for total_outcome in  range(1,19):
  count = 0
  for first_dice in range(1,7):
    for second_dice in range(1,7):
      for third_dice in range(1,7):

        if(first_dice + second_dice + third_dice == total_outcome):
          count += 1

  print(total_outcome, "=",round((count/216)*100,3 ))
```
